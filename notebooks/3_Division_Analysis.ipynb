```python
import pandas as pd

df = pd.read_csv("cleaned_nassau_candy_distributor.csv")

division_summary = df.groupby("Division").agg({
    "Sales": "sum",
    "Gross Profit": "sum",
    "Cost": "sum",
    "Units": "sum"
}).reset_index()

division_summary["Gross Margin %"] = (
    division_summary["Gross Profit"] /
    division_summary["Sales"]
) * 100

division_summary
```

```python
division_summary.sort_values(
    by="Gross Profit",
    ascending=False
)
```

```python
import matplotlib.pyplot as plt

division_summary.plot(
    x="Division",
    y="Gross Profit",
    kind="bar",
    figsize=(8, 5)
)

plt.title("Profit by Division")
plt.show()
```

```python
division_summary["Sales Contribution %"] = (
    division_summary["Sales"] /
    division_summary["Sales"].sum()
) * 100

division_summary["Profit Contribution %"] = (
    division_summary["Gross Profit"] /
    division_summary["Gross Profit"].sum()
) * 100

division_summary
```
